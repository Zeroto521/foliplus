# Demo

This notebook demonstrates all **foliplus** controls on a single map. Run each cell to see the components in action.


## 1. Import Dependencies

```python
import random

import folium
from folium.plugins import MarkerCluster

from foliplus import (
    ZH,
    Fullscreen,
    HeatmapControl,
    LayerControl,
    MapSearch,
    MeasureControl,
    ScaleControl,
)
```

## 2. Create Map & Sample Data

```python
m = folium.Map(location=[26.08, 119.30], zoom_start=12)

folium.TileLayer("OpenStreetMap", name="OpenStreetMap").add_to(m)
folium.TileLayer(
    "https://{s}.tile.opentopomap.org/{z}/{x}/{y}.png",
    name="Topographic",
    attr="© OpenTopoMap contributors",
).add_to(m)

# 50 random markers for heatmap / layer demo
mc = MarkerCluster(name="Sample Points").add_to(m)
for i in range(50):
    lat = 26.08 + random.uniform(-0.05, 0.05)
    lng = 119.30 + random.uniform(-0.05, 0.05)
    folium.Marker(
        [lat, lng],
        popup=f"Point {i}",
        icon=folium.Icon(color=random.choice(["red", "blue", "green"])),
    ).add_to(mc)

print("✅ Map + 50 markers ready")
```

## 3. Add All foliplus Controls

```python
# MapSearch — Chinese locale
MapSearch(zoom=16, position="topright", locale=ZH).add_to(m)

# LayerControl — enhanced layer manager
LayerControl().add_to(m)

# Fullscreen — clean fullscreen view
Fullscreen().add_to(m)

# ScaleControl — scale bar with zoom label
ScaleControl().add_to(m)

# HeatmapControl — H3 hexbin aggregation
HeatmapControl().add_to(m)

# MeasureControl — distance / circle / locate tools
MeasureControl().add_to(m)

print("✅ All 6 controls added")
```
